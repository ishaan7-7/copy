```python
text = """
# HELP test metric
good_metric 123
bad_metric abc
another_good 45
"""

def fake_parse(text):
    out = {}
    for line in text.splitlines():
        if not line or line.startswith("#"):
            continue
        try:
            name, value = line.split(" ", 1)
            out[name] = float(value)
        except Exception:
            continue
    return out

fake_parse(text)
```
